Функция потерь:

```python
def log_loss(w, x, y):
    # Отступ M = y * (w0*1 + w1*x1 + w2*x2)
    margin = y * (x @ w)
    # log2(z) = ln(z) / ln(2)
    return np.sum(np.log(1 + np.exp(-margin)) / np.log(2))
```

Производная функции потерь (градиент):

```python
def get_gradient(w, x, y):
    # Отступ M = y * (w0*1 + w1*x1 + w2*x2)
    margin = y * (x @ w)
    multiplier = -y / ((1 + np.exp(margin)) / np.log(2))
    grad = x.T @ multiplier
    return grad
```

Обучение:

```python
def train(x, y, eps=1e-4, iterations=10000):
    w = np.zeros(x.shape[1])
    lrs = np.array([0.1, 0.01, 0.01])

    for i in range(iterations):
        grad = get_gradient(w, x, y)

        # критерий остановки
        if np.linalg.norm(grad) < eps:
            print(f"Остановлено на итерации {i}, ||grad|| = {np.linalg.norm(grad):.6f}")
            break

        w = w - lrs * grad

        if i % 100 == 0:
            print(f"Итерация {i}: L = {log_loss(w, x, y):.4f}")
    return w
```

```python
weights = train(x, y)
w0, w1, w2 = weights
print("-" * 100)
print(f"Финальные веса w0, w1, w2: {weights}")
```
